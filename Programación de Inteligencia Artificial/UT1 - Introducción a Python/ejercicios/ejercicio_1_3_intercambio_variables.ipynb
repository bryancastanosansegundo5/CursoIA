#### Ejercicio 1.3 — Intercambio de variables
Dadas dos variables `a = 10` y `b = 20`, intercambia sus valores sin usar una tercera variable.

```python
# Pista: a, b = b, a
```

````python
a=10
b=20
print(f'Antes del cambio: a = {a}, b = {b}')

a,b =b,a
print(f'Despues del cambio: a = {a}, b = {b}')
````
